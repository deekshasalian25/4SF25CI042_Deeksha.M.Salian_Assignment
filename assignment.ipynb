DataFrame and product column

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

df['Product'] = df['A'] * df['B']

print("\nDataFrame after adding Product:")
print(df)
```

Row-wise and column-wise sum

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),
                  columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("\nRow-wise Sum:")
print(df.sum(axis=1))

print("\nColumn-wise Sum:")
print(df.sum(axis=0))
```

Nan and filling Nan with mean

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[3, 1] = np.nan

print("DataFrame with NaN:")
print(df)

df = df.fillna(df.mean())

print("\nDataFrame after filling NaN:")
print(df)
```

Drop rows containing NaN values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan

print("Original DataFrame:")
print(df)

df = df.dropna()

print("\nDataFrame after dropping NaN rows:")
print(df)
```

Data aggregation- group by category

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 100, 10)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nAggregation Result:")
print(result)
```

Group by Category and calculate total sales

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': np.random.choice(['Laptop', 'Phone', 'Tablet'], 8),
    'Category': np.random.choice(['Electronics', 'Accessories'], 8),
    'Sales': np.random.randint(100, 1000, 8)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales by Category:")
print(result)
```

Merging DataFrames

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Riya', 'John']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 78]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

result = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(result)
```

MultiIndex DataFrame – Indexing and Slicing

```python
import pandas as pd

arrays = [
    ['A', 'A', 'B', 'B'],
    [1, 2, 1, 2]
]

index = pd.MultiIndex.from_arrays(arrays, names=['Category', 'Number'])

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nAccessing Category A:")
print(df.loc['A'])

print("\nAccessing a particular element:")
print(df.loc[('B', 1)])

print("\nSlicing:")
print(df.loc['A':'B'])
```

MultiIndex with Category and SubCategory

```python
import pandas as pd
import numpy as np

index = pd.MultiIndex.from_product(
    [['A', 'B'], ['X', 'Y']],
    names=['Category', 'SubCategory']
)

df = pd.DataFrame({
    'Value': np.random.randint(1, 100, 4)
}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nSum of Values:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

Apply a function to double DataFrame values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2)

print("\nDoubled DataFrame:")
print(result)
```

Lambda function to calculate row-wise sum

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print("\nDataFrame with Sum column:")
print(df)
```

Convert Pandas Series strings to uppercase

```python
import pandas as pd

s = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

print("Original Series:")
print(s)

result = s.str.upper()

print("\nUppercase Series:")
print(result)
```
